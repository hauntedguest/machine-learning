<h2 align="center"> Support Vector Machine</h2>

### Problem Statement:  Classify raisins into one of the two categories,
1. Kecimen
1. Besni

```python
import pandas as pd

df = pd.read_excel("Raisin_Dataset.xlsx")
df.sample(5)
```

```python
df.shape
```

There are total 900 records and using all the features that we have available, we will build a classification model by using support vector machine 

### Train Test Split

```python
X = df[["Area", "MajorAxisLength", "MinorAxisLength", "Eccentricity", "ConvexArea", "Extent", "Perimeter"]]
y = df["Class"]

from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=10)
```

### Model Training Using SVM: RBF Kernel - No Scaling

```python
from sklearn.svm import SVC

model = SVC(kernel="rbf")
model.fit(X_train, y_train)

y_pred = model.predict(X_test)

from sklearn.metrics import classification_report

report = classification_report(y_test, y_pred)
print(report)

model.n_iter_
```

### Model Training Using SVM: Linear Kernel

```python
from sklearn.svm import SVC

model = SVC(kernel="linear")
model.fit(X_train, y_train)

y_pred = model.predict(X_test)

from sklearn.metrics import classification_report

report = classification_report(y_test, y_pred)
print(report)

model.n_iter_
```

### Scale the Data

```python
from sklearn.preprocessing import StandardScaler,MinMaxScaler
```

```python
scaler = StandardScaler()
scaler.fit(X_train)                      # learn mean/std from train only

X_train_scaled = scaler.transform(X_train)
X_test_scaled  = scaler.transform(X_test)   # reuse train's mean/std
```

```python
X_train_scaled
```

### Model Training Using SVM: RBF Kernel With Scaling

```python
from sklearn.svm import SVC

model = SVC(kernel="rbf")
model.fit(X_train_scaled, y_train)

y_pred = model.predict(X_test_scaled)

from sklearn.metrics import classification_report

report = classification_report(y_test, y_pred)
print(report)

model.n_iter_
```

### Model Training Using SVM: Linear Kernel With Scaling

```python
from sklearn.svm import SVC

model = SVC(kernel="linear")
model.fit(X_train_scaled, y_train)

y_pred = model.predict(X_test_scaled)

from sklearn.metrics import classification_report

report = classification_report(y_test, y_pred)
print(report)

model.n_iter_
```

### Making use of Sklearn Pipeline

```python
from sklearn.pipeline import Pipeline
```

```python
pipeline =Pipeline([
    ('scaler', StandardScaler()),
    ('svm', SVC(kernel="rbf"))
])
```

```python
pipeline.fit(X_train, y_train)
y_pred= pipeline.predict(X_test)

from sklearn.metrics import classification_report

report = classification_report(y_test, y_pred)
print(report)   

print(model.n_iter_)
```
